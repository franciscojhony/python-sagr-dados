```python
import pandas as pd
# Exibir todas as colunas sem truncamento
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

```python
df = pd.read_csv("escolas_relatorio.csv", sep=";") # Se o arquivo usar ponto e vírgula:

# exibir quantidade de linhas e colunas - df.shpae[0] pega o número de linhas e df.shpae[1] o de colunas.
print(f"Linhas: {df.shape[0]} | Colunas: {df.shape[1]}\n")
print(df.columns.tolist(), "\n") # exibe o nome das colunas em uma lista comum

# Filtrar por AnoLetivo = 2026 E SituacaoEducacenso = 'ATIVA'
df = df[(df["AnoLetivo"] == 2026) & (df["SituacaoEducacenso"] == "ATIVA")]
print(f"Linhas: {df.shape[0]} | Colunas: {df.shape[1]}\n")
display(df)
#df.head(3)
```

```python
#df = df[["Cod_Lotação", "CNPJ"]]
#display(df)
#df.to_excel("lotacao-cnpj.xlsx", sheet_name="lotacao", index=False)
```

```python
# Buscar CNPJ 07518670000143
cnpj_busca = "07518670000143"

# 1) Busca exata (só funciona se o CNPJ estiver completo no arquivo)
cnpj_limpo = df["CNPJ"].astype(str).str.replace(r"\D", "", regex=True).str.zfill(14)
resultado = df[cnpj_limpo == cnpj_busca]

# 2) Se não achar, busca aproximada (o CSV tem CNPJ em notação científica, ex.: 7,51867E+12)
if resultado.empty:
    cnpj_num = pd.to_numeric(df["CNPJ"].astype(str).str.replace(",", "."), errors="coerce")
    alvo = int(cnpj_busca)
    resultado = df[(cnpj_num - alvo).abs() < 1e8]  # tolerância pela perda de dígitos
    print("Busca aproximada (CNPJ truncado no arquivo):")

display(resultado)
```
