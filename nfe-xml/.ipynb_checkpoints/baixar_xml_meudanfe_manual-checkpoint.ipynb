# Download manual (assistido) de XML de NF-e pelo Meu Danfe

Versão **sem API**. O script só ajuda: abre o site, cola a chave, e **você** resolve a verificação anti-robô (Cloudflare), clica em **BUSCAR** e em **Baixar XML**. Depois o script confere se o arquivo chegou, renomeia para `{chave}.xml` e guarda em `xmls_baixados`.

Para o site não bloquear:
- baixe **poucas notas por vez**: defina `LINHA_INICIAL` e `LINHA_FINAL`, rode, e depois mude a faixa para o próximo lote;
- o script espera um tempo entre as notas (`PAUSA_ENTRE_NOTAS`);
- chaves já baixadas são puladas, então é só rodar de novo mais tarde.

**Modos:**
- `"chrome_aberto"` (recomendado): o script se conecta ao seu Chrome normal e **preenche a chave sozinho**. Você só passa pelo anti-robô, clica em BUSCAR e em Baixar XML.
- `"selenium"`: o script abre uma janela própria do Chrome (mais fácil de configurar, mas o Cloudflare pode reprovar).
- `"navegador_padrao"`: abre o site no seu navegador, sem controle; a chave só é mostrada na tela.

## 1. Bibliotecas

````python
import os
import time
import shutil
import random
import webbrowser
import pandas as pd

try:
    import pyperclip          # pip install pyperclip
except ImportError:
    pyperclip = None
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
````

## 2. Configuração

````python
# --- CONFIGURAÇÕES ---
ARQUIVO_EXCEL = "notas_fiscais.xlsx"
COLUNA_DA_CHAVE = "CHAVE_ACESSO_NOTA"
PASTA_DESTINO = "xmls_baixados"

# Faixa de linhas da planilha a processar (números de linha como aparecem no Excel:
# linha 1 = cabeçalho, primeira nota = linha 2). Após cada lote, altere os dois valores.
LINHA_INICIAL = 2
LINHA_FINAL = 6

MODO = "chrome_aberto"             # "chrome_aberto" (recomendado), "selenium" ou "navegador_padrao"
PORTA_DEBUG = 9222                 # porta usada pelo Chrome aberto com --remote-debugging-port
PAUSA_ENTRE_NOTAS = (20, 40)       # segundos de espera aleatória entre uma nota e outra
TEMPO_MAX_ESPERA = 180             # segundos para você concluir o download de cada nota
PASTA_DOWNLOADS_NAVEGADOR = os.path.join(os.path.expanduser("~"), "Downloads")  # usada no modo navegador_padrao
AVISO_ACIMA_DE = 10                # pede confirmação se a faixa tiver mais notas que isso (evita bloqueio)

destino = os.path.abspath(PASTA_DESTINO)
os.makedirs(destino, exist_ok=True)

df = pd.read_excel(ARQUIVO_EXCEL)
if COLUNA_DA_CHAVE not in df.columns:
    raise ValueError(f"Coluna '{COLUNA_DA_CHAVE}' não encontrada. Colunas: {list(df.columns)}")
if LINHA_INICIAL < 2 or LINHA_FINAL < LINHA_INICIAL:
    raise ValueError("Use LINHA_INICIAL >= 2 e LINHA_FINAL >= LINHA_INICIAL.")

total_linhas = len(df) + 1   # última linha de dados no Excel
fim_real = min(LINHA_FINAL, total_linhas)
faixa = df.iloc[LINHA_INICIAL - 2 : fim_real - 1]   # linha Excel N = índice N-2

lote, invalidas, repetidas = [], [], []
for idx, valor in faixa[COLUNA_DA_CHAVE].items():
    if pd.isna(valor):
        continue
    chave = str(valor).strip()
    if len(chave) != 44:
        invalidas.append((idx + 2, chave))
    elif chave in lote:
        repetidas.append((idx + 2, chave))
    else:
        lote.append(chave)

ja_baixadas = [c for c in lote if os.path.exists(os.path.join(destino, f"{c}.xml"))]
pendentes = [c for c in lote if c not in ja_baixadas]
lote = pendentes

print(f"Planilha: {total_linhas - 1} notas (linhas 2 a {total_linhas}). Faixa pedida: {LINHA_INICIAL} a {LINHA_FINAL}"
      + (f" (ajustada para {fim_real})" if fim_real != LINHA_FINAL else ""))
print(f"Na faixa: {len(pendentes)} a baixar | {len(ja_baixadas)} já baixadas | {len(invalidas)} inválidas | {len(repetidas)} repetidas")
for lin, ch in invalidas:
    print(f"   linha {lin}: chave com tamanho incorreto ({ch})")
if len(lote) > AVISO_ACIMA_DE:
    if input(f"Faixa com {len(lote)} notas (acima de {AVISO_ACIMA_DE}). Risco de bloqueio. Continuar? (s/n): ").strip().lower() != "s":
        raise SystemExit("Cancelado. Reduza a faixa.")
````

## 3. Funções de apoio

````python
def xmls_em(pasta):
    return {f: os.path.getmtime(os.path.join(pasta, f))
            for f in os.listdir(pasta) if f.lower().endswith(".xml")}

def baixando_ainda(pasta):
    return any(f.lower().endswith((".crdownload", ".tmp", ".part")) for f in os.listdir(pasta))

def esperar_novo_xml(pasta, antes, chave, limite):
    """Espera aparecer um .xml novo na pasta e devolve o caminho (ou None)."""
    fim = time.time() + limite
    while time.time() < fim:
        agora = xmls_em(pasta)
        novos = [f for f in agora if f not in antes or agora[f] > antes[f]]
        if novos and not baixando_ainda(pasta):
            com_chave = [f for f in novos if chave in f]
            escolhido = (com_chave or sorted(novos, key=lambda f: agora[f]))[-1]
            return os.path.join(pasta, escolhido)
        time.sleep(1)
    return None

def xml_contem_chave(caminho, chave):
    try:
        with open(caminho, "r", encoding="utf-8", errors="ignore") as f:
            return chave in f.read()
    except OSError:
        return False

def guardar(origem, chave):
    alvo = os.path.join(destino, f"{chave}.xml")
    if os.path.abspath(origem) != os.path.abspath(alvo):
        shutil.move(origem, alvo)
    return alvo
````

## 4. Baixar (assistido)

Para cada nota: a chave é colocada no campo (ou copiada), **você** passa pelo anti-robô, clica em **BUSCAR** e depois em **Baixar XML**. O script detecta o arquivo sozinho. No prompt: Enter = aguardar o download, `p` = pular a nota, `s` = parar tudo.

## Antes de rodar: abrir o Chrome no modo "chrome_aberto"

Para o script preencher a chave sozinho **no seu Chrome normal** (que costuma passar no anti-robô), abra o Chrome assim, com todas as janelas dele fechadas antes. No Prompt de Comando do Windows:

```
"C:\Program Files\Google\Chrome\Application\chrome.exe" --remote-debugging-port=9222 --user-data-dir="%USERPROFILE%\chrome_meudanfe"
```

Ele abre uma janela de Chrome com um perfil separado (a primeira vez, faça login no Meu Danfe se quiser). Deixe essa janela aberta e rode as células. Os downloads vão para a sua pasta Downloads, e o script move cada XML para `xmls_baixados`.

````python
driver = None
pasta_observada = PASTA_DOWNLOADS_NAVEGADOR

if MODO in ("chrome_aberto", "selenium"):
    from selenium import webdriver
    from selenium.webdriver.common.by import By
    from selenium.webdriver.chrome.service import Service
    from webdriver_manager.chrome import ChromeDriverManager
    opts = webdriver.ChromeOptions()
    if MODO == "chrome_aberto":
        # conecta ao Chrome que VOCÊ abriu com --remote-debugging-port (veja a célula de instruções)
        opts.add_experimental_option("debuggerAddress", f"127.0.0.1:{PORTA_DEBUG}")
        driver = webdriver.Chrome(options=opts)
    else:
        pasta_observada = os.path.join(destino, "_temp")
        os.makedirs(pasta_observada, exist_ok=True)
        opts.add_experimental_option("prefs", {
            "download.default_directory": pasta_observada,
            "download.prompt_for_download": False,
            "safebrowsing.enabled": True,
        })
        driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=opts)

def preencher_chave(chave):
    """Abre o site e escreve a chave no campo automaticamente."""
    driver.get("https://meudanfe.com.br/")
    campo = WebDriverWait(driver, 120).until(    # tempo para você passar pelo anti-robô, se pedir
        EC.presence_of_element_located((By.ID, "searchTxt")))
    try:
        campo.clear(); campo.send_keys(chave)
    except Exception:
        driver.execute_script(
            "arguments[0].value = arguments[1];"
            "arguments[0].dispatchEvent(new Event('input', {bubbles:true}));", campo, chave)

ok, puladas, falhas = [], [], []
try:
    for i, chave in enumerate(lote, 1):
        print(f"\n[{i}/{len(lote)}] Chave: {chave}")
        antes = xmls_em(pasta_observada)

        if driver:
            try:
                preencher_chave(chave)
                print("   Chave preenchida no campo automaticamente.")
            except Exception as e:
                print(f"   (não consegui preencher: {type(e).__name__}; digite a chave no site):", chave)
        else:
            print("   Modo sem controle do navegador: abra o site e digite a chave:", chave)
            webbrowser.open("https://meudanfe.com.br/")

        print("   >> Resolva o anti-robô se pedir, clique em BUSCAR e depois em 'Baixar XML'.")
        resp = input("   Enter = aguardar o download | p = pular | s = parar tudo: ").strip().lower()
        if resp == "s":
            break
        if resp == "p":
            puladas.append(chave); continue

        arq = esperar_novo_xml(pasta_observada, antes, chave, TEMPO_MAX_ESPERA)
        if not arq:
            falhas.append(chave)
            print(f"   ✗ Nenhum XML novo em {TEMPO_MAX_ESPERA}s. Rode de novo depois para tentar essa chave.")
            continue
        if not xml_contem_chave(arq, chave):
            print("   ⚠ O XML baixado não contém essa chave. Conferir manualmente:", arq)
            falhas.append(chave)
            continue
        guardar(arq, chave)
        ok.append(chave)
        print(f"   ✓ Salvo como {chave}.xml")

        if i < len(lote):
            espera = random.randint(*PAUSA_ENTRE_NOTAS)
            print(f"   Aguardando {espera}s antes da próxima (evita bloqueio)...")
            time.sleep(espera)
finally:
    if driver and MODO == "selenium":
        driver.quit()   # no modo chrome_aberto o seu Chrome continua aberto

print(f"\nResumo: {len(ok)} baixadas | {len(puladas)} puladas | {len(falhas)} falhas")
print(f"Ainda pendentes nesta faixa: {len(pendentes) - len(ok)}")
print(f"Próxima faixa sugerida: LINHA_INICIAL = {fim_real + 1}")
print(f"Pasta: {destino}")
````
